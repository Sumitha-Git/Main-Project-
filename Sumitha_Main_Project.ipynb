**Dataset Column/features Description:**

* 1.	EV_Charging_Station_Name -	Name of the EV charging station.

* 11.	Charging_Mode -	Charging mode or method used during the session.
* 12.	Charger_Power	- Maximum power capacity of the charger, generally measured in kW.
* 13.	Connector_Type	- Type/standard of connector used to connect the EV to the charger.
* 14.	Connector_Code	- Specific code or identifier representing the connector.
* 15.	Currency -	Currency used for calculating the charging cost, e.g., INR, USD.
* 16.	Price_per_kWh	- Charging price for consuming one kWh of electricity.
* 17.	Session_Start_Datetime -	Date and time when the charging session started.
* 18.	Session_End_Datetime -	Date and time when the charging session ended.
* 19.	Duration -	Total duration of the charging session, usually represented as time.
* 20.	Duration_Minutes	- Charging session duration converted into minutes. Useful for analysis and visualization.
* 21.	Energy_Qty_kWh	- Total electrical energy delivered to the EV during the session, measured in kWh.
* 22.	Vehicle_Make_Model	- Manufacturer and model of the EV.
* 23.	Vehicle_Type	 - Specific type/model classification of the vehicle.
* 24.	Vehicle_Battery_kWh	- Battery capacity of the vehicle, measured in kWh.
* 25.	SOC_Start_%	 - State of Charge (SOC) of the battery at the beginning of charging, expressed as a percentage.
* 26.	SOC_End_%	 - State of Charge of the battery at the end of charging, expressed as a percentage.
* 27.	Payment_Method	- Method used to pay for the charging session, such as UPI, card, wallet, etc.
* 28.	Payment_Status	- Indicates whether payment was successful, pending, failed, refunded, etc.
* 29.	Session_Energy_Amount	- Amount charged for the electricity/energy consumed during the session.
* 30.	Session_Fee	 - Additional fee charged for the charging session, excluding energy cost where applicable.
* 31.	Untaxed_Amount	- Charging amount before taxes are applied.
* 32.	Tax_Rate_%	- Tax rate applied to the charging transaction, expressed as a percentage.
* 33.	Tax_Amount	- Total tax amount applied to the charging transaction.
* 34.	Session_Total	- Final total amount payable for the charging session, including applicable fees and taxes.
* 35.	Meter_Start_kWh	- Electricity meter reading at the beginning of the charging session.
* 36.	Meter_End_kWh	- Electricity meter reading at the end of the charging session.
* 37.	Session_Status -	Status of the charging session, such as completed, active, cancelled, failed, etc.
* 38.	Idle_Time_Minutes	- Time in minutes that the vehicle remains connected to the charger after charging or during an idle period.
* 39.	Tariff_Type	- Type of pricing/tariff plan applied to the charging session, such as fixed, variable, time-based, etc.

```python
#Handle missing values
# Price_Per_kWh is a numarical column, Here I have to find the Outliers using bleow code
# I found Min outliere is 0.00
# I Found Max outliers is 0.56

df['Price_per_kWh'].describe()
```

```python
#IQR method for Price_per_kWh
#Find Q1 → the lower quarter of prices and Q3 - the the upper quarter of prices
#Calculate the IQR =Q3-Q1
#Calculate the lower and upper limits

#Calculate Q1 and Q3
Q1 = df["Price_per_kWh"].quantile(0.25)
Q3 = df["Price_per_kWh"].quantile(0.75)

#Calculate IQR
IQR = Q3 - Q1

#Calculate the lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

#Display the calculated values
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

#Create a new cleaned dataset
df_cleaned = df[
    (df["Price_per_kWh"] >= lower_limit) &
    (df["Price_per_kWh"] <= upper_limit)
]

#Compare row counts
print(f"Original row count: {len(df)}")
print(f"Cleaned row count: {len(df_cleaned)}")
print(f"Outliers removed: {len(df) - len(df_cleaned)}")
```

```python

plt.figure(figsize=(10, 5))

plt.boxplot(df["Price_per_kWh"].dropna(), vert=False)

plt.xlabel("Price per kWh")
plt.title("Box Plot of Price_per_kWh")
plt.grid(axis="x")

plt.show()
```

```python
# Charger_Power is the Categorical.Because this column contains different categories
# Filling missing Charger_Power using the "mode"

mode_power = df['Charger_Power'].mode()[0]
df['Charger_Power'] = df['Charger_Power'].fillna(mode_power)


print("Null values remaining:", df['Charger_Power'].isnull().sum())
```

```python
# Connector_Type is the Categorical.Because this column contains different categories
# Filling missing Charger_Power using the "mode"

mode_connector = df['Connector_Type'].mode()[0]
df['Connector_Type'] = df['Connector_Type'].fillna(mode_connector)

print("Null values remaining:", df['Connector_Type'].isnull().sum())
```

```python
# Skewness Values in my Dataset

skewness = df.skew(numeric_only=True)
print("Skewness for numeric columns:")
print(skewness)
```

```python
#This scatterplot shows the relationship between Charging Duration (Minutes) and Energy Delivered (kWh) across vehicle categories.
#Session duration alone does not determine energy delivered—fast chargers deliver massive energy within 90 minutes, while slow chargers cap out below 60 kWh even when plugged in for 3 hours.

evc.set_theme(style="whitegrid")

plt.figure(figsize=(7, 5))
evc.scatterplot(data=df, x='Duration_Minutes', y='Energy_Qty_kWh', hue='Vehicle_Category', alpha=0.6, palette='Set2')
plt.title('Duration vs. Energy Delivered', fontsize=12, fontweight='bold')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Energy (kWh)')
plt.tight_layout()
plt.show()
```

```python
#This comparative boxplot analyzes charging session durations across Electric 2-Wheelers and Electric 4-Wheelers.
#Vehicle category has no impact on session duration—both 2-wheelers and 4-wheelers occupy charging bays for the exact same amount of time on average.

plt.figure(figsize=(7, 5))
evc.boxplot(
    data=df,
    x='Vehicle_Category',
    y='Duration_Minutes',
    hue='Vehicle_Category',
    palette='Set2',
    legend=False,
)

plt.title('Duration Minutes by Vehicle Category', fontsize=12, fontweight='bold')
plt.xlabel('Vehicle Category')
plt.ylabel('Duration (Minutes)')
plt.tight_layout()
plt.show()
```

```python
#Faster Energy Delivery: DC chargers transfer much more energy in less time (up to 350 kWh in 100 mins) compared to slower AC chargers (under 65 kWh in 190 mins).
#Higher Revenue from DC: Total session cost increases directly with energy used, but DC chargers generate significantly higher revenue per session ($29.35 average) than AC chargers ($5.85 average).
#Independent Idle Time: Driver idle time after charging shows no connection to session length or energy delivered across either charger type.

evc.set_theme(style="whitegrid")

pairplot_vars = [
    "Duration_Minutes",
    "Energy_Qty_kWh",
    "Session_Total",
    "Idle_Time_Minutes",
]

g = evc.pairplot(
    df[pairplot_vars + ["Charger_Type"]],
    hue="Charger_Type",
    palette="Set1",
    corner=True,
)

g.fig.suptitle(
    "Multivartity Pairplot of Charging Session Metrics by Charger Type", y=1.02, fontsize=8
)

plt.show()
```

```python
#Consistent Energy Usage: Vehicle energy consumption (IQR) stays nearly identical across all three tariff types for both 2-Wheelers and 4-Wheelers.
#High-Energy Outliers: Fast DC charging creates massive energy spikes (100 to 350 kWh) across every single tariff option.
#No Charging Limits: Drivers draw high energy levels regardless of which tariff plan they use.

evc.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

evc.boxplot(
    data=df,
    x="Vehicle_Category",
    y="Energy_Qty_kWh",
    hue="Tariff_Type",
    palette="Set2",
)


plt.title("Energy Delivered (kWh) across Vehicle Categories grouped by Tariff Type")
plt.xlabel("Vehicle Category")
plt.ylabel("Energy Delivered (kWh)")
plt.xticks(rotation=15)
plt.legend(title="Tariff Type")


plt.tight_layout()
plt.show()
```

```python
#Higher 4-Wheeler Revenue: Electric 4-Wheelers generate slightly higher average revenue per session than 2-Wheelers in almost every state.
#Key Regional Highs & Lows: Telangana leads in 2-Wheeler revenue (11.51), while Puducherry records the lowest 4-Wheeler revenue (10.27).
#Consistent Billing: Across major states like Tamil Nadu and Karnataka, session costs stay very balanced between both vehicle types (10.68 to 11.16).

evc.set_theme(style="whitegrid")
pivot_df = df.pivot_table(
    index="Vehicle_Category", columns="State", values="Session_Total", aggfunc="mean"
)

plt.figure(figsize=(8, 6))

evc.heatmap(
    pivot_df,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    cbar_kws={"label": "Mean Session Total"},
)

plt.title(
    "Pivot Table Heatmap: Mean Session Total by Vehicle Category & State"
)
plt.xlabel("State")
plt.ylabel("Vehicle_Category")

plt.tight_layout()
plt.show()
```

Predictive Analysis (What is likely to happen?)
* •	More City Demand: Big tech cities like Bengaluru will see higher charger use as more people buy 4-Wheelers.
* •	Power Grid Stress: Large fleet vehicles using over 200 kWh will cause sudden power spikes on local grids.
* •	Blocked Chargers: Drivers staying plugged in for over 90 minutes will cause long waits for others looking for open chargers.
          Prescriptive Analysis (What should we do?)
* •	Add More Fast Chargers: Put high-speed DC chargers in busy cities for 4-Wheelers.
* •	Charge Overstay Fees: Add extra fees when sessions go over 90 minutes so drivers do not hog chargers.
* •	Make Payments Easy: Focus on quick payments using UPI, wallets, mobile apps, and card readers.
# **Conclusions**
* •	Focus on Fast Charging: Expanding DC fast chargers in major cities will support the growing number of 4-Wheeler users.
* •	Reduce Station Waiting: Charging extra fees for sessions over 90 minutes will keep chargers open for new drivers.
* •	Protect the Power Grid: Planning for sudden high-energy draws over 200 kWh will prevent local grid overloads.
* •	Streamline User Payments: Keeping app, UPI, and card payments fast will improve driver experience at all stations.
